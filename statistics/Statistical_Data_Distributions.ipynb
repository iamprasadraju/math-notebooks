### Distribution

A **distribution** describes how the values of a variable are distributed across their possible values.

> **What values can occur, and how much probability or frequency is associated with each value?**

So, a distribution describes the **pattern of probability or frequency across values**.

### 1. Empirical distribution

A distribution constructed **directly from observed sample data**.

It describes how the values in the observed sample are distributed.

For a discrete variable, we can express it using **empirical probabilities**:

$$
\hat P(X=x)=\frac{\text{number of observations equal to }x}{n}
$$

These are based on observed data rather than a theoretical probability model.

### 2. Theoretical / Probability distribution

A **mathematical model that assigns probabilities to the possible outcomes of a random variable**.

For example, for a fair die:

$$
P(X=x)=\frac{1}{6}, \qquad x\in\{1,2,3,4,5,6\}
$$

The probabilities come from the probability model, rather than from the frequencies observed in a particular sample.

### Probability and Statistics

A useful high-level distinction is:

> **Probability: model → possible data**

Start with a probability model and reason about what data could occur.

> **Statistics: observed data → inference about model**

Start with observed data and use it to learn about the underlying population or probability model.

**One nuance:** Statistics does not only work with empirical distributions. Statistical inference often uses **theoretical probability distributions** to model uncertainty and make conclusions from observed data.


```python
# simulation of rolling a dice - Probability Experiment
import random
import numpy as np

num_exp = 1000
observations = []

for i in range(num_exp):
    out = random.choice([i for i in range(1, 7)])
    observations.append(out)

# print(observations)
```

```python
# Emperical Distribution
from collections import Counter
counts = Counter(observations)

for value in range(1, 7):
    print(value, ":", counts[value], "P =", counts[value] / num_exp)


import matplotlib.pyplot as plt

plt.hist(sorted(sample_space))
plt.show()
```
