```python
# Bernoulli Distribution (only two outputs: success or failure)
# probability of success = p
# probability of failure = 1 - p

import random

samples = []
for _ in range(100):
    samples.append(random.choice([0, 1]))

print(samples)
```

```python
from collections import Counter

count = Counter(samples)

# H : 0, T: 1
print(f"P(X = H): {count[0] / 100}%, P(X = T): {count[1] / 100}%")

import matplotlib.pyplot as plt

plt.hist(samples)
plt.show()
```
